```python
import re

pattern = r'^(Ground|Lower Basement|Upper Basement|\d+)\s+out of\s+\d+$'

df["floor"] = df["floor"].where(df["floor"].astype(str).str.match(pattern))
```

```python
df["floor"].value_counts(dropna=False)
```

```python
df = df.dropna(subset=["floor"])
```
